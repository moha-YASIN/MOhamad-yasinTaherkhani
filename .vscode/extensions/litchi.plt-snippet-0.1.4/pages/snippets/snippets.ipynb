## 绘图

```python
# {"name": "误差折线图", "title": "带有误差信息的折线图，支持上下误差不同"}
fig, ax = plt.subplots(constrained_layout=True)

# yerr,xerr为上下和左右误差，通常只指定一个
# capsize,capthick为误差棒长宽，ecolor为误差棒颜色
# 更多参数：https://www.wolai.com/matplotlib/6QuMVNfhxsMMnMp7VSpkUk
ax.errorbar(x, y, yerr, marker="s", capsize=4, capthick=2, ecolor="k", label="y")

# 设置坐标轴标签
ax.set_xlabel("X (unit)", fontweight="bold")
ax.set_ylabel("Y (unit)", fontweight="bold")

# 添加图例
_ = ax.legend()

plt.show()
```

```python
# {"name": "柱状图", "title": "创建柱状图、条形图"}
fig, ax = plt.subplots(constrained_layout=True)

# bar_labels是一个字符串列表，bar_values是一个与bar_labels等长的数字列表
# 更多参数：https://www.wolai.com/matplotlib/2mypc3HyBTyLR8TQKmxVku
bars = ax.bar(bar_labels, bar_values, color="salmon")

# 为条形添加数字标签, padding是标签与条形的距离
ax.bar_label(bars, padding=3)

# 设置坐标轴标签
ax.set_xlabel("X (unit)", fontweight="bold")
ax.set_ylabel("Y (unit)", fontweight="bold")

# 添加图例
_ = ax.legend()

plt.show()
```

```python
# {"name": "饼图", "title": "展示数据中各成分的比例"}
fig, ax = plt.subplots(constrained_layout=True)

# x: 饼图中各成分比例
x = [0.3, 0.1, 0.1]
# labels: 饼图中各成分标签
labels = ["label 1", "label 2", "label 3"]

# autopct: 饼图中各成分标签的显示格式
# 更多参数：https://www.wolai.com/matplotlib/rKrwDAh6rTo7mRKpSTTopz
_ = ax.pie(x, labels=labels, autopct="%1.1f%%")

plt.show()
```

```python
# {"name": "散点图", "title": "创建散点图"}
fig, ax = plt.subplots(constrained_layout=True)

# 关键字s指定散点的大小，更多参数：https://www.wolai.com/matplotlib/6QQ2tpDuAS7ggBMNP8YXbK
_ = ax.scatter(x, y, s=24, label="label")

# 设置坐标轴标签
ax.set_xlabel("X (unit)", fontweight="bold")
ax.set_ylabel("Y (unit)", fontweight="bold")

# 添加图例
_ = ax.legend()

plt.show()
```

```python
# {"name": "像素图", "title": "imshow函数，也可用于热度图，要求输入一个二维数组"}
fig, ax = plt.subplots(constrained_layout=True)

# 更多参数：https://www.wolai.com/matplotlib/9S3VvdWev3tNckFvKcMHKA
im = ax.imshow(array2d, cmap="coolwarm")

# 添加颜色条，更多参数：https://www.wolai.com/matplotlib/3JLQ4xkFa9az9ed5w9FHt4
cbar = fig.colorbar(im, ax=ax, pad=0.05, shrink=1.00)
cbar.set_label("label", fontweight="bold")

plt.show()
```

```python
# {"name": "等高线图", "title": "contour和contourf函数，要求输入一个二维数组"}
fig, ax = plt.subplots(constrained_layout=True)
# 设置坐标轴比例
ax.set_aspect(aspect=1.0)

# contour 生成等高线，contourf 填充等高线之间的区域，可单独使用也可以配合使用（levels参数相同）
# 更多参数：https://www.wolai.com/matplotlib/5Re8LXrTNvjN5BxDcNFwwL
_ = ax.contour(
    array2d, colors="gray", alpha=0.6, levels=5, linewidths=2, linestyles="-"
)
cs = ax.contourf(array2d, levels=5, cmap="coolwarm")

# 绘制颜色条
cbar = fig.colorbar(cs, ax=ax, shrink=0.75, aspect=15)
cbar.set_label("Z", fontweight="bold", rotation=0)

# 保存图片
# plt.savefig('./等高线填充图.png', dpi=300)

# 显示图片
plt.show()
```

## 轴与标签

```python
# {"name": "轴范围", "title": "手动设置X/Y轴的上下限"}
# 设置X/Y轴的上下限
ax.set_xlim([x_min, x_max])
ax.set_ylim([y_min, y_max])
```

```python
# {"name": "轴标签", "title": "设置X/Y轴的标签"}
# x, y: 标签沿X轴或Y轴的位置，默认为 0.5
# labelpad: 浮点数关键字，控制标签与轴之间的距离
# fontsize: 文本字体大小,  rotation: 文本旋转角度
ax.set_xlabel("X (unit)", fontweight="bold")
ax.set_ylabel("Y (unit)", fontweight="bold")
```

```python
# {"name": "轴刻度", "title": "设置X/Y轴的刻度位置"}
ax.set_xticks([0, 1, 2, 3])
ax.set_yticks([-1, 0, 1, 2, 3])
```

```python
# {"name": "刻度标签", "title": "设置X/Y轴的刻度标签"}
# 注意保证 ticklabels 与 ticks 长度一致
ax.set_xticklabels([0, 1, 2, 3], color="k")
ax.set_yticklabels(["a", "b", "c", "d", "e"], color="k")
```

```python
# {"name": "轴分布", "title": "设置轴刻度的分布，常用线性-linear、指数-log"}
# 线型-linear、指数-log、对称对数-symlog、对数概率-logit
ax.set_yscale("log")
```

```python
# {"name": "轴比例", "title": "设置Y/X轴比例"}
# aspect=dy/dx, dx,dy为一个像素的宽高对应的坐标轴数值
ax.set_aspect(aspect=0.5)
```

## 其它常用

```python
# {"name": "图例", "title": "添加图例"}
# ncol: 图例的列数，loc: 图例位置，"upper right", "upper left", "lower left", "center", ...
# reverse: False or True, 图例项顺序是否反转
# 更多参数详见 https://www.wolai.com/matplotlib/7Lx6EZgnxScUd2diem67W8
ax.legend()
```

```python
# {"name": "颜色条", "title": "添加colorbar"}
# mappable 是一个可绘制的对象，比如imshow, contourf返回的对象
# shrink 标识颜色条的长度，aspect 标识颜色条的高度，pad 标识颜色条与绘图区的距离
# 更多参数详见 https://www.wolai.com/matplotlib/3JLQ4xkFa9az9ed5w9FHt4
cbar = fig.colorbar(mappable, shrink=0.8, aspect=12, pad=0.04, location="right")

# 设置颜色条的标签， y 表示颜色条的位置，labelpad 表示标签与颜色条的浮动距离
cbar.set_label("m/s", y=0.5, labelpad=-2)
```

```python
# {"name": "保存", "title": "保存图片，后缀支持 png,jpg,pdf,eps等"}
# transparent: False or True, 透明背景
plt.savefig("filename.png", dpi=300)
plt.savefig("filename.eps")
```

```python
# {"name": "网格", "title": "添加背景网格"}
# axis='y'|'x'|'both'
# which='minor'|'major'|'both'
# ls = 实线'-'|虚线'--'|点画线'-.'|点线':'
ax.grid(True, axis="both", ls="--", color="gray", alpha=0.6)
```

```python
# {"name": "文本", "title": "添加文本信息"}
# verticalalignment: 文本竖直方向锚点位置，可选'center'、'bottom'、'top', 简写va
# horizontalalignment: 文本水平方向锚点位置，可选'center'、'left'、'right', 简写ha
# 更多参数：https://www.wolai.com/matplotlib/tQWP8MznCduRZzjDkdUyxh
ax.text(0.5, 0.5, "text here", fontsize=10)
```

```python
# {"name": "注解", "title": "添加注释，相当于 文本+箭头"}
# xy是箭头指向的位置，xytext是文本的位置
# 更多参数：https://www.wolai.com/matplotlib/oAj9m2iuBGJSy7osNtaXmr
ax.annotate(
    "注解文本内容",
    xy=(0.1, 0.7),
    xytext=(0.6, 0.8),
    arrowprops={
        "arrowstyle": "-|>",
        "linewidth": 2,
        "facecolor": "k",
    },
)
```

```python
# {"name": "3D坐标", "title": "创建3D坐标系"}
# 创建3D坐标系
fig, ax = plt.subplots(figsize=(3, 2), subplot_kw={"projection": "3d"})
```

```python
# {"name": "多图布局", "title": "同一个图中并存多个坐标系"}
fig, axs = plt.subplots(
    nrows=2,  # 行数
    ncols=2,  # 列数
    figsize=(6, 6),  # 画布大小(英寸)
    gridspec_kw={
        "height_ratios": [0.5, 0.5],  # 行高比例
        "width_ratios": [0.5, 0.5],  # 列宽比例
        "hspace": 0.05,  # 行间距(英寸)
        "wspace": 0.25,  # 列间距(英寸)
    },
)
```

```python
# {"name": "区间填充", "title": "在指定范围内填充颜色"}
# 在x范围内填充0和y之间的区域，color为填充颜色，alpha为透明度
# 更多参数：https://www.wolai.com/matplotlib/3QEbYtS7pZttmVUq2wTgZ7
_ = ax.fill_between(x, 0, y, color="C0", alpha=0.5)
```

```python
# {"name": "方块填充", "title": "在一段区间内设置背景颜色"}
# axhspan和hlines为水平方向, axhspan和hlines为垂直方向，参数也需要调整为ymin, ymax
# 更多参数：https://www.wolai.com/matplotlib/4sGGJFdbmXmdNrW9UDiRu5
ax.axhspan(0.7, 0.9, color="lightgray", alpha=0.8)
```
